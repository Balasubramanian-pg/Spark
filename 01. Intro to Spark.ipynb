# Introduction to Apache Spark

Welcome to this comprehensive introduction to **Apache Spark**, the lightning-fast, distributed cluster-computing framework designed for big data processing and analytics.


## 1. What is Apache Spark?

Apache Spark is an open-source, distributed processing system used for big data workloads. It utilizes in-memory caching and optimized query execution for fast analytic queries against data of any size.

### Why Spark over Traditional MapReduce?
* **In-Memory Processing:** Unlike Hadoop MapReduce, which writes intermediate results to disk between operations, Spark keeps data in memory across multiple stages of a computation. This can make Spark up to **100 times faster** for iterative algorithms.
* **Ease of Use:** Spark provides native APIs in Scala, Java, Python (PySpark), and R, allowing developers to write applications quickly.
* **Unified Engine:** Spark integrates SQL queries, streaming data, machine learning (MLlib), and graph processing (GraphX) into a single unified framework.

## 2. Core Architecture of Spark

Spark follows a **Master-Worker** architecture consisting of a driver program and worker nodes managed by a cluster manager.

* **Driver Program:** The process that runs the `main()` function of the application, creates the `SparkSession`, and converts the user code into tasks distributed across the cluster.
* **Cluster Manager:** Allocates resources across the cluster (e.g., Apache Mesos, Hadoop YARN, Kubernetes, or Spark's built-in Standalone manager).
* **Executors:** Worker nodes launched on the cluster that execute the actual tasks assigned by the driver and store data in memory or disk storage.

## 3. Core Abstractions: RDDs vs. DataFrames

Understanding how Spark represents data is crucial for writing efficient applications.

### A. Resilient Distributed Datasets (RDD)
* The fundamental building block of Spark.
* A fault-tolerant collection of operational elements that can be processed in parallel.
* **Lazy Evaluation:** Transformations on RDDs are not computed immediately. Spark remembers the lineage of transformations and only evaluates them when an *action* is triggered.

### B. DataFrames
* Modeled after pandas and R DataFrames, Spark DataFrames organize data into named columns (like a relational table).
* Built on top of RDDs, they include an **Optimized Execution Engine (Catalyst Optimizer)** that drastically improves performance over raw RDDs.

## 4. Setting up Spark in Python (PySpark)

Let's initialize our environment by creating a `SparkSession`, which is the entry point to programming Spark with the Dataset and DataFrame API.

```python
# Import PySpark and initialize SparkSession
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("IntroToSparkNotebook") \
    .getOrCreate()

# Verify the Spark session
spark
```

## 5. Creating and Exploring a DataFrame

Now, let's create a sample DataFrame from local data to understand basic transformations (`select`, `filter`) and actions (`show`, `collect`).

```python
# Sample Data: Employee Records
data = [
    ("Alice", "Engineering", 85000),
    ("Bob", "Marketing", 62000),
    ("Charlie", "Engineering", 92000),
    ("Diana", "HR", 55000),
    ("Evan", "Marketing", 70000)
    ]

columns = ["Name", "Department", "Salary"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Action: Show contents
df.show()
```

### Performing Transformations and Actions

* **Transformations** (e.g., `filter`, `select`, `groupBy`) return a *new* DataFrame and are lazy.
* **Actions** (e.g., `show`, `count`, `collect`) trigger the actual execution pipeline.

```python
# Transformation: Filter employees with salary > 65,000
high_earners = df.filter(df["Salary"] > 65000)

# Action: Display results
high_earners.show()

# Aggregation: Average salary by department
from pyspark.sql.functions import avg
df.groupBy("Department").agg(avg("Salary").alias("Avg_Salary")).show()
```
